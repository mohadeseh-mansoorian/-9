```python

# ===== ۱. موتور استنتاج (مغز دکتر) =====
def doctor_diagnose(name, age, has_fever, has_cough):
    diagnosis = ""
    advice = []

    # تشخیص بر اساس علائم (منطق قوانین اگر-آنگاه)
    if has_fever and has_cough:
        diagnosis = "آنفولانزا یا عفونت تنفسی"
        advice = [
            "استراحت کامل در خانه و پرهیز از تماس با دیگران",
            "نوشیدن مایعات فراوان (آب، آبمیوه، سوپ گرم)",
            "استفاده از داروهای تب‌بر (طبق دستور پزشک)"
        ]
    elif has_fever and not has_cough:
        diagnosis = "عفونت احتمالی یا التهاب"
        advice = [
            "بهتر است برای معاینه دقیق‌تر و آزمایش خون به پزشک مراجعه کنید.",
            "فعلاً استراحت کنید و مایعات بنوشید."
        ]
    elif not has_fever and has_cough:
        diagnosis = "سرماخوردگی ساده یا آلرژی"
        advice = [
            "استفاده از شربت سرفه یا آنتی‌هیستامین",
            "غرغره کردن آب نمک ولرم برای تسکین گلو",
            "دوری از هوای سرد و آلوده"
        ]
    else:
        diagnosis = "شما سالم هستید!"
        advice = ["علائم خاصی از بیماری ندارید. مراقب سلامت خود باشید."]

    # توصیه‌های خاص بر اساس سن (هوشمندی سیستم)
    if age < 12:
        advice.append("⚠️ توجه: چون شما در سن کودکی هستید، حتماً والدین خود را در جریان بگذارید.")
    elif age > 65:
        advice.append("⚠️ توجه: در سنین بالا سیستم ایمنی ضعیف‌تر است. اگر علائم تشدید شد، حتماً به اورژانس بروید.")

    return diagnosis, advice

# ===== ۲. بخش تعامل با بیمار (رابط کاربری) =====
print("👨‍⚕️ سلام! من دکتر هوش مصنوعی هستم. لطفاً به سوالات من پاسخ دهید تا شما را معاینه کنم.\n")

# گرفتن اسم
name = input("۱. لطفاً نام خود را وارد کنید: ")

# گرفتن سن
while True:
    try:
        age = int(input(f"۲. {name} عزیز، چند سالتان است؟ "))
        break
    except ValueError:
        print("لطفاً سن را به صورت عدد وارد کنید.")

# گرفتن علائم
print(f"\nبسیار خب {name} جان، حالا به این سوالات با بله (yes) یا خیر (no) پاسخ دهید:")
fever_input = input("۳. آیا تب دارید؟ (yes/no): ").lower()
cough_input = input("۴. آیا سرفه می‌کنید؟ (yes/no): ").lower()

# تبدیل جواب‌ها به True/False
has_fever = (fever_input in ["yes", "y", "بله", "ب"])
has_cough = (cough_input in ["yes", "y", "بله", "ب"])

# ===== ۳. گرفتن نتیجه و چاپ نسخه =====
print("\n" + "="*40)
print("⏳ در حال تحلیل اطلاعات توسط هوش مصنوعی...")
print("="*40 + "\n")

diagnosis, advice = doctor_diagnose(name, age, has_fever, has_cough)

print(f"🩺 تشخیص دکتر: {diagnosis}")
print(f"\n📋 نسخه و توصیه‌های پزشکی برای {name} عزیز:")
for i, item in enumerate(advice, 1):
    print(f"  {i}. {item}")
print("\nشفای عاجل! اگر علائم ادامه داشت حتماً به پزشک متخصص مراجعه کنید. 🌹")
```
