```python

city_map = {
    "Tehran": ["Qazvin", "Karaj"],
    "Qazvin": ["Zanjan"],
    "Karaj": ["Hamedan"],
    "Zanjan": ["Hamedan"],
}

def dfs_count(graph, start, goal, path=None, visited=None):
    # مقداردهی اولیه
    if path is None:
        path = [start]
    if visited is None:
        visited = {start}  # اصلاح باگ پی‌دی‌اف (استفاده از آکولاد برای ساخت مجموعه)

    if start == goal:
        return path, len(visited)  # برگرداندن مسیر و تعداد گره‌های بازدیدشده

    for neighbor in graph.get(start, []):
        if neighbor not in visited:
            visited.add(neighbor)
            result, count = dfs_count(graph, neighbor, goal, path + [neighbor], visited)
            if result:
                return result, count
    return None, len(visited)

# اجرای کد
path, visited_count = dfs_count(city_map, "Tehran", "Hamedan")
print("مسیر پیدا شده:", path)
print("تعداد گره‌های بازدیدشده:", visited_count)
```
