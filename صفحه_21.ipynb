```python

# ===== کدهای صفحه ۲۰ =====
initial_state = { "farmer": False, "wolf": False, "sheep": False, "cabbage": False }
goal_state = { "farmer": True, "wolf": True, "sheep": True, "cabbage": True }

def is_goal(state):
    return state == goal_state

print("Have we reached the goal?", is_goal(initial_state))


# ===== کدهای صفحه ۲۱ =====
def is_safe(state):
    if state["wolf"] == state["sheep"] != state["farmer"]:
        return False
    if state["sheep"] == state["cabbage"] != state["farmer"]:
        return False
    return True

def get_successors(state):
    successors = []
    items = ["wolf", "sheep", "cabbage", None]

    for item in items:
        new_state = dict(state)
        new_state["farmer"] = not state["farmer"]

        if item:
            new_state[item] = not state[item]

        if is_safe(new_state):
            successors.append(new_state)

    return successors

print("Successors:", get_successors(initial_state))
```
