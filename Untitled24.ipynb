```python
import heapq

# ===== ۱. الگوریتم Greedy Best-First Search =====
def greedy_bfs(graph, start, goal, h):
    # صف اولویت‌دار بر اساس h(n) مرتب می‌شود
    frontier = [(h[start], 0, [start])] # (h, cost, path)
    visited = set()

    while frontier:
        # همیشه گرهی که کمترین h را دارد، اول خارج می‌شود
        f, g, path = heapq.heappop(frontier)
        node = path[-1]

        if node == goal:
            return path, g

        if node in visited:
            continue
        visited.add(node)

        for neighbor, cost in graph.get(node, []):
            if neighbor not in visited:
                new_g = g + cost
                # در Greedy، اولویت فقط بر اساس h است
                heapq.heappush(frontier, (h[neighbor], new_g, path + [neighbor]))
    return None, None


# ===== ۲. الگوریتم A* =====
def a_star(graph, start, goal, h):
    # صف اولویت‌دار بر اساس f(n) = g(n) + h(n) مرتب می‌شود
    frontier = [(h[start], 0, [start])]
    visited = set()

    while frontier:
        f, g, path = heapq.heappop(frontier)
        node = path[-1]

        if node == goal:
            return path, g

        if node in visited:
            continue
        visited.add(node)

        for neighbor, cost in graph.get(node, []):
            new_g = g + cost
            new_f = new_g + h[neighbor] # ترکیب هزینه واقعی و تخمینی
            heapq.heappush(frontier, (new_f, new_g, path + [neighbor]))
    return None, None


# ===== ۳. تست و مقایسه =====
# تعریف یک گراف که در آن Greedy اشتباه می‌کند
graph = {
    "Start": [("A", 100), ("B", 200)],
    "A": [("Goal", 100)],
    "B": [("Goal", 300)],
    "Goal": []
}

# تخمین فاصله تا هدف (Goal)
h = {"Start": 150, "A": 200, "B": 50, "Goal": 0}

greedy_path, greedy_cost = greedy_bfs(graph, "Start", "Goal", h)
astar_path, astar_cost = a_star(graph, "Start", "Goal", h)

print("--- خروجی Greedy Best-First Search ---")
print(f"مسیر: {greedy_path}")
print(f"هزینه کل: {greedy_cost}")

print("\n--- خروجی الگوریتم A* ---")
print(f"مسیر: {astar_path}")
print(f"هزینه کل: {astar_cost}")
```
