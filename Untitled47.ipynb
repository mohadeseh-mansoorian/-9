```python
# ============================================================
# سامانه تحلیل هوشمند مجموعه داده گل زنبق (IRIS)
# منبع: UCI Machine Learning Repository
# Dataset ID: 53
# کتابخانه اصلی: Pandas
# زبان گزارش: فارسی
# ============================================================

!pip -q install -U ucimlrepo

import pandas as pd
import numpy as np

from ucimlrepo import fetch_ucirepo
from IPython.display import display, Markdown

pd.set_option("display.max_columns", 30)
pd.set_option("display.max_colwidth", 100)
pd.set_option("display.width", 150)

# ------------------------------------------------------------
# توابع کمکی برای نمایش گزارش
# ------------------------------------------------------------

def title(text):
    display(Markdown(f"## {text}"))
    print("-" * 75)

def message(text):
    print("●", text)


# ------------------------------------------------------------
# ۱. دریافت دیتاست از مخزن UCI
# ------------------------------------------------------------

title("۱. اتصال به مخزن UCI و دریافت دیتاست")

try:
    iris = fetch_ucirepo(id=53)

    X = iris.data.features.copy()
    y = iris.data.targets.copy()

    df = pd.concat([X, y], axis=1)
    df = df.loc[:, ~df.columns.duplicated()]

    metadata = iris.metadata

    message("اتصال به مخزن UCI با موفقیت انجام شد.")
    message("دیتاست گل زنبق دریافت شد.")

except Exception as error:
    print("خطا در دریافت دیتاست:", error)
    raise


# ------------------------------------------------------------
# ۲. اطلاعات عمومی دیتاست
# ------------------------------------------------------------

title("۲. مشخصات عمومی مجموعه داده")

print("نام دیتاست:", metadata.name)
print("شناسه دیتاست:", metadata.uci_id)
print("حوزه موضوعی:", metadata.area)
print("تعداد نمونه ها:", metadata.num_instances)
print("تعداد ویژگی ها:", metadata.num_features)
print("نوع مسئله:", metadata.tasks)
print("وجود مقادیر خالی:", metadata.has_missing_values)
print("تعداد رکوردهای دریافت شده:", df.shape[0])
print("تعداد ستون های دریافت شده:", df.shape[1])

print("\nتوضیحات دیتاست:")
print(metadata.abstract)


# ------------------------------------------------------------
# ۳. نمایش نمونه رکوردها
# ------------------------------------------------------------

title("۳. مشاهده نمونه اطلاعات گل ها")

message("پنج رکورد اول دیتاست:")
display(df.head())

message("پنج رکورد تصادفی:")
display(df.sample(min(5, len(df)), random_state=42))


# ------------------------------------------------------------
# ۴. معرفی ویژگی های دیتاست
# ------------------------------------------------------------

title("۴. بررسی مشخصات ویژگی ها")

translation = {
    "sepal length": "طول کاسبرگ",
    "sepal width": "عرض کاسبرگ",
    "petal length": "طول گلبرگ",
    "petal width": "عرض گلبرگ",
    "class": "گونه گل"
}

variables = iris.variables.copy()

variables["توضیح فارسی"] = (
    variables["name"].map(translation).fillna("سایر ویژگی ها")
)

info_columns = [
    col for col in [
        "name", "توضیح فارسی", "role",
        "type", "description", "units",
        "missing_values"
    ]
    if col in variables.columns
]

display(variables[info_columns])


# ------------------------------------------------------------
# ۵. بررسی کیفیت و سلامت داده ها
# ------------------------------------------------------------

title("۵. بررسی کیفیت اطلاعات")

quality = pd.DataFrame({
    "نوع داده": df.dtypes.astype(str),
    "تعداد مقادیر معتبر": df.notna().sum(),
    "تعداد مقادیر خالی": df.isna().sum(),
    "درصد مقادیر خالی": (df.isna().mean() * 100).round(2),
    "تعداد مقادیر منحصربه فرد": df.nunique()
})

quality.index.name = "نام ستون"

display(quality)

missing_total = df.isna().sum().sum()
duplicate_total = df.duplicated().sum()

message(f"مجموع مقادیر خالی: {missing_total}")
message(f"تعداد رکوردهای تکراری: {duplicate_total}")


# ------------------------------------------------------------
# ۶. شناسایی ستون های غیرضروری
# ------------------------------------------------------------

title("۶. شناسایی ستون های کم ارزش")

target_columns = list(y.columns)

useless_columns = []

for col in df.columns:

    missing_ratio =df[col].isna().mean()
    unique_count = df[col].nunique(dropna=True)

    if col in target_columns:
        continue

    if missing_ratio > 0.80 or unique_count <= 1:
        useless_columns.append(col)

analysis_df = df.drop(columns=useless_columns).copy()

if useless_columns:
    print("ستون های شناسایی شده:")
    print(useless_columns)
else:
    message("تمام ستون های دیتاست دارای اطلاعات قابل استفاده هستند.")

print("تعداد ستون های اولیه:", df.shape[1])
print("تعداد ستون های قابل استفاده:", analysis_df.shape[1])


# ------------------------------------------------------------
# ۷. تحلیل آماری ویژگی های عددی
# ------------------------------------------------------------

title("۷. تحلیل آماری ویژگی های عددی")

numeric_df = analysis_df.select_dtypes(include=np.number)

numeric_report = numeric_df.describe().T

numeric_report = numeric_report.rename(columns={
    "count": "تعداد",
    "mean": "میانگین",
    "std": "انحراف معیار",
    "min": "کمترین مقدار",
    "25%": "چارک اول",
    "50%": "میانه",
    "75%": "چارک سوم",
    "max": "بیشترین مقدار"
})

numeric_report.index.name = "نام ویژگی"

display(numeric_report.round(2))


# ------------------------------------------------------------
# ۸. بررسی توزیع گونه های گل
# ------------------------------------------------------------

title("۸. بررسی تعداد و توزیع گونه های گل")

target_name = y.columns[0]

target_report = (
    df[target_name]
    .value_counts(dropna=False)
    .to_frame("تعداد نمونه")
)

target_report["درصد از کل"] = (
    target_report["تعداد نمونه"] / len(df) * 100
).round(2)

target_report.index.name = "نام گونه"

display(target_report)

print("تعداد گونه های مختلف:", df[target_name].nunique())


# ------------------------------------------------------------
# ۹. مقایسه ویژگی های سه گونه گل
# ------------------------------------------------------------

title("۹. مقایسه ویژگی های گونه های مختلف")

feature_columns = list(X.columns)

grouped_report = (
    analysis_df
    .groupby(target_name)[feature_columns]
    .agg(["mean", "min", "max"])
)

grouped_report = grouped_report.round(2)

display(grouped_report)

print("\nمیانگین ویژگی ها در هر گونه:")

mean_report = (
    analysis_df
    .groupby(target_name)[feature_columns]
    .mean()
    .round(2)
)

display(mean_report)


# ------------------------------------------------------------
# ۱۰. شناسایی مقادیر پرت
# ------------------------------------------------------------

title("۱۰. شناسایی مقادیر پرت")

outlier_results = []

for col in numeric_df.columns:

    values = numeric_df[col].dropna()

    if len(values) < 4 or values.nunique() < 2:
        continue

    q1 = values.quantile(0.25)
    q3 = values.quantile(0.75)

    iqr = q3 - q1

    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr

    outliers = values[
        (values < lower) | (values > upper)
    ]

    outlier_results.append({
        "ویژگی": translation.get(col, col),
        "تعداد مقادیر پرت": len(outliers),
        "درصد مقادیر پرت": round(
            len(outliers) / len(values) * 100, 2
        ),
        "حد پایین": round(lower, 2),
        "حد بالا": round(upper, 2)
    })

if outlier_results:
    display(pd.DataFrame(outlier_results))
else:
    message("مقدار پرت قابل بررسی شناسایی نشد.")


# ------------------------------------------------------------
# ۱۱. بررسی همبستگی ویژگی ها
# ------------------------------------------------------------

title("۱۱. بررسی ارتباط آماری ویژگی ها")

correlation_matrix = numeric_df.corr()

correlation_matrix.index = [
    translation.get(col, col)
    for col in correlation_matrix.index
]

correlation_matrix.columns = [
    translation.get(col, col)
    for col in correlation_matrix.columns
]

display(correlation_matrix.round(3))

message("اعداد نزدیک به مثبت یک نشان دهنده رابطه خطی مثبت قوی هستند.")
message("اعداد نزدیک به منفی یک نشان دهنده رابطه خطی منفی قوی هستند.")


# ------------------------------------------------------------
# ۱۲. شناسایی ویژگی های مهم در تفکیک گونه ها
# ------------------------------------------------------------

title("۱۲. بررسی میزان تفاوت ویژگی ها بین گونهها")

comparison = []

for col in feature_columns:

    group_means = (
        analysis_df
        .groupby(target_name)[col]
        .mean()
    )

    difference = group_means.max() - group_means.min()

    comparison.append({
        "ویژگی": translation.get(col, col),
        "بیشترین میانگین": round(group_means.max(), 2),
        "کمترین میانگین": round(group_means.min(), 2),
        "اختلاف میانگین ها": round(difference, 2)
    })

comparison_df = pd.DataFrame(comparison)

comparison_df = comparison_df.sort_values(
    "اختلاف میانگین ها",
    ascending=False
)

display(comparison_df)


# ------------------------------------------------------------
# ۱۳. گزارش نهایی
# ------------------------------------------------------------

title("۱۳. گزارش نهایی تحلیل دیتاست")

summary = {
    "نام مجموعه داده": metadata.name,
    "شناسه مخزن": metadata.uci_id,
    "تعداد کل نمونه ها": df.shape[0],
    "تعداد ویژگی های ورودی": X.shape[1],
    "تعداد متغیرهای هدف": y.shape[1],
    "تعداد گونه های گل": df[target_name].nunique(),
    "تعداد ستون های قابل استفاده": analysis_df.shape[1],
    "تعداد ستون های کم ارزش": len(useless_columns),
    "تعداد ویژگی های عددی": numeric_df.shape[1],
    "تعداد مقادیر خالی": missing_total,
    "تعداد رکوردهای تکراری": duplicate_total
}

summary_df = pd.DataFrame(
    list(summary.items()),
    columns=["عنوان گزارش", "نتیجه"]
)

display(summary_df)

print("\nپیش نمایش دیتاست نهایی:")
display(analysis_df.head(10))

print("\nتحلیل مجموعه داده گل زنبق با موفقیت به پایان رسید.")
print("=" * 75)
```
