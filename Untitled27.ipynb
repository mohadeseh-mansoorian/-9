```python

# ۱. نصب کتابخانه‌های لازم
!pip install scikit-learn pandas openpyxl -q

import os
import struct
import pandas as pd
import numpy as np
from sklearn.tree import DecisionTreeClassifier
from datetime import datetime
from google.colab import files

# ===== ۲. ساخت یک فایل JKS تستی برای اینکه کد چیزی برای پیدا کردن داشته باشد =====
with open("/content/my_test_key.jks", "wb") as f:
    f.write(b'\xfe\xed\xfe\xed') # امضای واقعی JKS
    f.write(b'\x00' * 5000)      # حجم الکی

# ===== ۳. آماده‌سازی هوش مصنوعی (آموزش مدل) =====
# به هوش مصنوعی یاد می‌دهیم که چه فایلی JKS است و چه فایلی نیست
# ویژگی‌ها: [حجم فایل به کیلوبایت، عدد ۴ بایت اول]
X_train = [
    [10, 4277010157],  # نمونه JKS (0xFEEDFEED = 4277010157)
    [500, 4277010157], # نمونه JKS
    [2, 1196314761],   # نمونه عکس PNG
    [1500, 1179603533],# نمونه فایل PDF
    [50, 0]            # فایل متنی خالی
]
# برچسب‌ها: 1 یعنی JKS، 0 یعنی فایل دیگر
y_train = [1, 1, 0, 0, 0]

# ساخت مدل هوش مصنوعی (درخت تصمیم‌گیری) و آموزش آن
ai_model = DecisionTreeClassifier()
ai_model.fit(X_train, y_train)

# ===== ۴. تابعی برای استخراج ویژگی‌های فایل =====
def get_file_features(filepath):
    try:
        size_kb = os.path.getsize(filepath) / 1024
        with open(filepath, 'rb') as f:
            magic_bytes = f.read(4)
            if len(magic_bytes) < 4:
                magic_number = 0
            else:
                magic_number = struct.unpack('>I', magic_bytes)[0]
        return size_kb, magic_number
    except:
        return 0, 0

# ===== ۵. اسکن پوشه‌ها با هوش مصنوعی =====
def scan_with_ai(search_dir):
    jks_list = []
    print(f"هوش مصنوعی در حال اسکن مسیر: {search_dir} ...")

    for root, dirs, files_list in os.walk(search_dir):
        for file in files_list:
            filepath = os.path.join(root, file)

            # استخراج ویژگی‌ها
            size, magic = get_file_features(filepath)

            # پیش‌بینی با هوش مصنوعی (آیا این فایل JKS است؟)
            prediction = ai_model.predict([[size, magic]])[0]

            if prediction == 1:
                mod_time = datetime.fromtimestamp(os.path.getmtime(filepath)).strftime('%Y-%m-%d %H:%M:%S')
                jks_list.append({
                    "نام فایل": file,
                    "مسیر": filepath,
                    "حجم (KB)": round(size, 2),
                    "تاریخ": mod_time
                })
                print(f"🤖 هوش مصنوعی تشخیص داد: {file} یک فایل JKS است.")
    return jks_list

# ===== ۶. اجرای اصلی و ساخت اکسل =====
SEARCH_PATH = "/content"
found_files = scan_with_ai(SEARCH_PATH)

if len(found_files) > 0:
    df = pd.DataFrame(found_files)
    excel_filename = "AI_JKS_Report.xlsx"
    df.to_excel(excel_filename, index=False)

    print(f"\n✅ هوش مصنوعی {len(found_files)} فایل JKS پیدا کرد.")
    print("📥 در حال دانلود فایل اکسل...")
    files.download(excel_filename)
else:
    print("\n❌ هوش مصنوعی هیچ فایل JKS پیدا نکرد.")
```
