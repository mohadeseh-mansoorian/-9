```python
import heapq

def a_star(graph, start, goal, h):
    frontier = [(h[start], 0, [start])]
    visited = set()

    while frontier:
        f, g, path = heapq.heappop(frontier)
        node = path[-1]

        if node == goal:
            return path, g

        if node in visited:
            continue
        visited.add(node)

        for neighbor, cost in graph.get(node, []):
            new_g = g + cost
            new_f = new_g + h[neighbor]
            heapq.heappush(frontier, (new_f, new_g, path + [neighbor]))
    return None, None

# ===== نقشه شهرهای ایران (فاصله‌ها به کیلومتر تقریبی) =====
# هدف ما رسیدن به "Kerman" است.
city_map = {
    "Tehran": [("Isfahan", 500), ("Shiraz", 900)],
    "Isfahan": [("Tehran", 500), ("Kerman", 600)],
    "Shiraz": [("Tehran", 900), ("Kerman", 400)],
    "Kerman": []
}

# تخمین فاصله هر شهر تا کرمان (Heuristic)
heuristic_to_kerman = {
    "Tehran": 800,
    "Isfahan": 500,
    "Shiraz": 300,
    "Kerman": 0
}

# پیدا کردن کوتاه‌ترین مسیر از تهران به کرمان
path, cost = a_star(city_map, "Tehran", "Kerman", heuristic_to_kerman)

print("--- نتیجه جستجو با A* ---")
print(f"مسیر پیدا شده: {path}")
print(f"هزینه کل (کیلومتر): {cost}")
```
