```python

def print_farmer_path():
    # تعریف مسیر دستی حل معما
    # False یعنی سمت چپ رودخانه، True یعنی سمت راست
    path = [
        {"farmer": False, "wolf": False, "sheep": False, "cabbage": False}, # حالت اولیه
        {"farmer": True,  "wolf": False, "sheep": True,  "cabbage": False}, # بردن گوسفند
        {"farmer": False, "wolf": False, "sheep": True,  "cabbage": False}, # برگشتن
        {"farmer": True,  "wolf": True,  "sheep": True,  "cabbage": False}, # بردن گرگ
        {"farmer": False, "wolf": True,  "sheep": False, "cabbage": False}, # برگرداندن گوسفند
        {"farmer": True,  "wolf": True,  "sheep": False, "cabbage": True},  # بردن کلم
        {"farmer": False, "wolf": True,  "sheep": False, "cabbage": True},  # برگشتن
        {"farmer": True,  "wolf": True,  "sheep": True,  "cabbage": True}   # بردن گوسفند (هدف)
    ]

    # توضیحات متناظر با هر گام
    explanations = [
        "حالت اولیه (همه سمت چپ)",
        "بردن گوسفند به سمت راست",
        "برگشتن کشاورز به سمت چپ",
        "بردن گرگ به سمت راست",
        "برگرداندن گوسفند به سمت چپ",
        "بردن کلم به سمت راست",
        "برگشتن کشاورز به سمت چپ",
        "بردن گوسفند به سمت راست (حالت هدف)"
    ]

    print("--- مسیر حرکت کشاورز ---")
    for i in range(len(path)):
        print(f"گام {i}: {explanations[i]}")
        print(f"وضعیت: {path[i]}")
        print("-" * 30)

# فراخوانی تابع
print_farmer_path()
```
