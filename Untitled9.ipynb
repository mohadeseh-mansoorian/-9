```python
# ===== کدهای صفحه ۲۰ =====
# Representing each state of the Farmer-Wolf-Sheep-Cabbage problem as a dictionary.
# True means "across the river".

initial_state = { "farmer": False, "wolf": False, "sheep": False, "cabbage": False }
goal_state = { "farmer": True, "wolf": True, "sheep": True, "cabbage": True }

def is_goal(state):
    return state == goal_state

print("Have we reached the goal?", is_goal(initial_state))
```
